## Independent and dependent features

```python
dataset.head(2)
```

```python
x = dataset.iloc[:,:-1]
y = dataset.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size =0.3, random_state=19)
```

Standard scalar: Our main goal in linear regresion is that we reach near or to global minimum. all the different features in our datasets are calculated based on different units and we use gradeint descent to reach gobal minimum point but if points are of different units it will take a lot of time to reach that. so thats why we scale the data before we use it

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_train=scaler.fit_transform(x_train)
```

```python
# we just use the fit instead of fit_transform for test data, bcoz we just want to apply the same transformation that we did for train dataset, to make sure thatmy model doesnot know much about my test dat
x_test = scaler.transform(x_test)
```

```python
print(x_test)
```

## model training

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
## print coefficients and the other is intercept
lr.coef_
```

```python
print(lr.intercept_)
```

```python
## on which parameters model has been trained
lr.get_params()
```

```python
## Prediction with test data
reg_pred=lr.predict(x_test)
```

```python
reg_pred
```

```python
## plot a scatter plot for pred
plt.scatter(reg_pred,y_test)
```

```python
residuals = reg_pred - y_test # error btw pred and y_test
```

```python
## plot this residuals

sns.displot(residuals,kind='kde')
```

## performance metrics

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

print(mean_absolute_error(reg_pred,y_test))
print(mean_squared_error(reg_pred,y_test))
print(np.sqrt(mean_squared_error(reg_pred,y_test)))
```

## r square and adjusted r square


```python
from sklearn.metrics import r2_score
score = r2_score(y_test, reg_pred)
print(score)
```
